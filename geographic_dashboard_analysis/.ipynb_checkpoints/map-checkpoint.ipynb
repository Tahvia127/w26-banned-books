### Import modules

```python
import numpy as np
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt
import json
```

```python
import plotly.io as pio
pio.renderers.default = 'chrome'
```

### Import data

```python
us_states = json.load(open('US_States.json', 'r'))
```

```python
# read the data into a pandas dataframe
banned_books = pd.read_csv("PEN_Combined_21-25.csv")
banned_books.head()
```

```python
state_populations = pd.read_csv('./data/raw/Census/ACSDP5Y2023.DP05-Data.csv')
state_populations.head()
```

```python
#Isolate population from data
state_populations = state_populations[['NAME', 'DP05_0001E']].rename(columns={'NAME': 'State', 'DP05_0001E': 'Population'}).drop(0)
state_populations['Population'] = state_populations['Population'].astype(int)
state_populations.head()
```

### Clean up data

```python
state_bans = banned_books['State'].value_counts().to_frame('bans').reset_index()
state_bans[state_bans['State'] == 'Nation']
```

The PEN America dataset includes bans listed as "Nation". After doing some research, it seems that the Department of Defense Education Activity had banned 596 books in 2025. This is very close to the 590 bans listed in the table and since it isn't a state, I decided to remove it.

https://pen.org/books-banned-by-department-of-defense-schools/

```python
state_bans = state_bans.drop(state_bans[state_bans['State'] == 'Nation'].index, inplace=False)
```

```python
all_states = ["Alabama", "Alaska", "Arizona", "Arkansas", "California", "Colorado", "Connecticut", "Delaware", "Florida", "Georgia", "Hawaii", 
              "Idaho", "Illinois", "Indiana", "Iowa", "Kansas", "Kentucky", "Louisiana", "Maine", "Maryland", "Massachusetts", "Michigan", "Minnesota", 
              "Mississippi", "Missouri", "Montana", "Nebraska", "Nevada", "New Hampshire", "New Jersey", "New Mexico", "New York", "North Carolina", 
              "North Dakota", "Ohio", "Oklahoma", "Oregon", "Pennsylvania", "Rhode Island", "South Carolina", "South Dakota", "Tennessee", "Texas", "Utah", 
              "Vermont", "Virginia", "Washington", "West Virginia", "Wisconsin", "Wyoming"]
state_bans = state_bans.set_index('State').reindex(all_states).reset_index()
state_bans.head()
```

Added back in missing states for complete map view

```python
state_bans = pd.merge(state_populations, state_bans, how='inner', on='State')
```

### Generate choropleth map

```python
state_bans['scale'] = state_bans['bans'] / state_bans['Population'] * 1000000
state_bans[['bans' ,'scale']] = state_bans[['bans' ,'scale']].fillna(0)
```

Normalize book bans to every 100,000 residents. Fill in NaN values with 0. 

```python
state_id_map = {}
for feature in us_states['features']:
    feature['id'] = feature['properties']['GEO_ID']
    state_id_map[feature['properties']['NAME']] = feature['id']
```

```python
state_bans['id'] = state_bans['State'].apply(lambda x: state_id_map[x])
state_bans.head()
```

Map geojson ids to dataset

```python
fig = px.choropleth(state_bans, 
                    title='Banned Books by State',
                    labels={'scale':'Per 100,000 residents'},
                    color='scale',
                    locations='id', 
                    geojson=us_states, 
                    scope='usa', 
                    hover_name='State',
                    hover_data=['bans'],
                    locationmode='geojson-id',
                    )
fig.update_layout(coloraxis_colorbar_title_side="right")
fig.show()
```
