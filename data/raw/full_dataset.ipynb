```python
import pandas as pd
import numpy as np
```

```python
import re
from datetime import datetime

def normalize_month_year_safe(date_str):
    """Normalize month-year strings to YYYY-MM-DD"""
    if pd.isna(date_str):
        return None
    
    # Remove extra spaces, non-breaking spaces, and weird characters
    date_str = re.sub(r"[\u00A0]", " ", date_str)
    date_str = date_str.strip()
    
    # Try full month name
    try:
        dt = datetime.strptime(date_str, "%B %Y")
        return dt.strftime("%Y-%m-%d")
    except:
        pass
    
    # Try abbreviated month
    try:
        dt = datetime.strptime(date_str, "%b %Y")
        return dt.strftime("%Y-%m-%d")
    except:
        print(f"⚠ Could not parse date: {date_str}")
        return None

# Load data
csv_path = 'data_filtered.csv'
df = pd.read_csv(csv_path)
print(f"Loaded {len(df)} books")

# Normalize dates
df['Ban Date Normalized'] = df['Ban Date'].apply(normalize_month_year_safe)

# Clean data (remove rows with invalid dates)
df = df.dropna(subset=['Ban Date Normalized', 'Title', 'State'])
print(f"After cleaning: {len(df)} books with valid dates")

# Save cleaned version
df.to_csv('data_clean.csv', index=False)
print("✓ Saved cleaned data to 'data_clean.csv'")
print(df[['Title', 'Ban Date', 'Ban Date Normalized']].head())
```
