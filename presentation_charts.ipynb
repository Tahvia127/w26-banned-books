## 0. Setup & Configuration

```python
import pandas as pd
import numpy as np
import altair as alt
import os
from scipy import stats
from itertools import combinations
from datetime import datetime

# ── Color Palette ────────────────────────────────────────────
TERRACOTTA = "#BF5B45"
GOLD       = "#C4A83D"
CHARCOAL   = "#3D3D3D"
SAGE       = "#B5C4A3"
CLAY       = "#A57B64"

PALETTE_5 = [TERRACOTTA, GOLD, CHARCOAL, SAGE, CLAY]
BG_COLOR   = "#FAFAF7"
TEXT_COLOR  = CHARCOAL
GRID_COLOR = "#E0DDD5"

# ── Altair Theme ─────────────────────────────────────────────
alt.data_transformers.disable_max_rows()

def branded_theme():
    return {
        "config": {
            "background": BG_COLOR,
            "title": {"fontSize": 18, "font": "Helvetica Neue, Arial, sans-serif",
                      "color": TEXT_COLOR, "anchor": "start",
                      "subtitleFontSize": 13, "subtitleColor": CLAY},
            "axis":  {"labelFontSize": 12, "titleFontSize": 13,
                      "titleColor": TEXT_COLOR, "labelColor": TEXT_COLOR,
                      "gridColor": GRID_COLOR, "domainColor": TEXT_COLOR,
                      "tickColor": TEXT_COLOR},
            "legend": {"labelFontSize": 12, "titleFontSize": 13,
                       "labelColor": TEXT_COLOR, "titleColor": TEXT_COLOR},
            "view": {"strokeWidth": 0},
            "range": {"category": PALETTE_5},
        }
    }

alt.themes.register("banned_books", branded_theme)
alt.themes.enable("banned_books")

VIZ_DIR = "visualizations"
os.makedirs(VIZ_DIR, exist_ok=True)

def save_chart(chart, name, w=700, h=420):
    sized = chart.properties(width=w, height=h)
    path = os.path.join(VIZ_DIR, f"{name}.png")
    try:
        sized.save(path, scale_factor=2)
        print(f"✓ Saved {path}")
    except Exception as e:
        print(f"⚠ PNG save failed ({e}) — displaying inline only")
    return sized

print("✓ Setup complete")
```

## 1. Load & Process Data

### 1a. Load raw files

```python
# ── File paths — edit if yours differ ────────────────────────
PEN_PATH      = "PEN_Combined_21-25_Final.csv"
GT_PATH       = "my_google_trends_results.csv"
GR_PATH       = "goodreads_engagement_all.csv"
GR_IDS_PATH   = "goodreads_ids_all.csv"
CENSUS_PATH   = "ACSDP5Y2023.DP05-Data.csv"
ELECTION_PATH = "Presidential Election Results by US County 2024.xlsx"

pen_raw = pd.read_csv(PEN_PATH)
gt_raw  = pd.read_csv(GT_PATH)
gr_raw  = pd.read_csv(GR_PATH)
census_raw = pd.read_csv(CENSUS_PATH, skiprows=[1])
election_raw = pd.read_excel(ELECTION_PATH)

print(f"PEN America:     {len(pen_raw):>6,} rows  ({pen_raw['State'].nunique()} states, {pen_raw['District'].nunique()} districts)")
print(f"Google Trends:   {len(gt_raw):>6,} rows  ({gt_raw['book_title'].nunique()} titles)")
print(f"Goodreads:       {len(gr_raw):>6,} rows")
print(f"Census:          {len(census_raw):>6,} rows")
print(f"Election:        {len(election_raw):>6,} rows")
```

### 1b. Clean Google Trends

```python
# Keep only successful collections
gt = gt_raw[gt_raw["collection_successful"] == True].copy()
print(f"Successful GT collections: {len(gt)}")

# Rename absolute_change → spike_magnitude for consistency
gt = gt.rename(columns={"absolute_change": "spike_magnitude"})

# Compute derived columns
gt["searches_increased"] = gt["spike_magnitude"] > 0
gt["pct_finite"] = gt["percent_change"].replace([np.inf, -np.inf], np.nan)

# Pre-ban popularity tiers (based on avg_search_before)
def popularity_tier(val):
    if val >= 50:  return "Very Popular (50+)"
    if val >= 20:  return "Popular (20-50)"
    if val >= 5:   return "Moderate (5-20)"
    if val > 0:    return "Low Interest (<5)"
    return "Niche (0)"

gt["pre_ban_popularity"] = gt["avg_search_before"].apply(popularity_tier)

# Streisand category
def streisand_cat(row):
    before, after = row["avg_search_before"], row["avg_search_after"]
    pct = row["pct_finite"]
    if before == 0 and after > 0:   return "New Interest"
    if before == 0 and after == 0:  return "No Google Presence"
    if pd.isna(pct):                return "No Google Presence"
    if pct > 25:                    return "Strong Spike"
    if pct > 0:                     return "Mild Increase"
    if pct > -10:                   return "No Change"
    if pct > -50:                   return "Decline"
    return "Strong Decline"

gt["streisand_category"] = gt.apply(streisand_cat, axis=1)

# Add ban counts from PEN
ban_counts = pen_raw.groupby("Title").agg(
    total_ban_instances=("Title", "size"),
    states_banned_in=("State", "nunique"),
    districts_banned_in=("District", "nunique"),
).reset_index().rename(columns={"Title": "book_title"})

gt = gt.merge(ban_counts, on="book_title", how="left")

print(f"\nStreisand categories:")
print(gt["streisand_category"].value_counts().to_string())
print(f"\nPopularity tiers:")
print(gt["pre_ban_popularity"].value_counts().to_string())
```

### 1c. Build challenge reason categories from PEN Description field

```python
# PEN doesn't have a structured challenge_reasons column,
# so we classify from the Description text using keyword matching.

CATEGORY_KEYWORDS = {
    "LGBTQ+ identity & sexuality": [
        "lgbtq", "transgender", "homosexual", "gay", "lesbian", "bisexual",
        "queer", "nonbinary", "non-binary", "gender identity", "same-sex",
        "drag", "trans boy", "trans girl", "transitioning"
    ],
    "Race & systemic inequality": [
        "racial", "racism", "race relations", "systemic", "slavery",
        "civil rights", "Black lives", "police brutality", "segregation",
        "discrimination", "African American", "Indigenous", "colonialism"
    ],
    "Sexual or romantic content": [
        "sexual content", "sexually explicit", "sex scene", "nudity",
        "pornograph", "erotic", "sexual activity", "sex ed", "masturbat",
        "sexual intercourse"
    ],
    "Violence & mature themes": [
        "violence", "graphic violence", "murder", "abuse", "assault",
        "gore", "warfare", "torture", "domestic violence", "shooting"
    ],
    "Mental health & trauma": [
        "mental health", "suicide", "depression", "anxiety", "trauma",
        "self-harm", "eating disorder", "ptsd", "drug use", "addiction",
        "substance abuse"
    ],
    "Age-appropriateness": [
        "age-appropriate", "age appropriate", "too mature", "not suitable",
        "children should not", "inappropriate for", "profanity", "language",
        "vulgar", "obscen"
    ],
    "Gender politics & feminism": [
        "feminism", "feminist", "gender roles", "patriarchy", "women's rights",
        "misogyn", "sexism", "reproductive"
    ],
}

def classify_description(text):
    if pd.isna(text):
        return []
    text_lower = text.lower()
    matched = []
    for category, keywords in CATEGORY_KEYWORDS.items():
        if any(kw in text_lower for kw in keywords):
            matched.append(category)
    return matched

pen_raw["categories_list"] = pen_raw["Description"].apply(classify_description)
pen_raw["challenge_reasons"] = pen_raw["categories_list"].apply(lambda x: "; ".join(x) if x else "")
pen_raw["num_reasons"] = pen_raw["categories_list"].apply(len)

# Build title-level categories
title_cats = (
    pen_raw.groupby("Title")["categories_list"]
    .apply(lambda lists: list(set(cat for sublist in lists for cat in sublist)))
    .reset_index()
    .rename(columns={"Title": "book_title", "categories_list": "reasons_list"})
)
title_cats["challenge_reasons"] = title_cats["reasons_list"].apply(lambda x: "; ".join(sorted(x)))
title_cats["num_reasons"] = title_cats["reasons_list"].apply(len)

categorized = title_cats[title_cats["num_reasons"] > 0]
uncategorized = title_cats[title_cats["num_reasons"] == 0]
print(f"Titles categorized: {len(categorized)} ({len(categorized)/len(title_cats)*100:.1f}%)")
print(f"Titles uncategorized: {len(uncategorized)} ({len(uncategorized)/len(title_cats)*100:.1f}%)")

# Merge into GT
cats_gt = gt.merge(title_cats[["book_title", "challenge_reasons", "reasons_list"]], on="book_title", how="left")
cats_gt["reasons_list"] = cats_gt["reasons_list"].apply(lambda x: x if isinstance(x, list) else [])
print(f"\nGT titles with categories: {(cats_gt['challenge_reasons'] != '').sum()} / {len(cats_gt)}")
```

### 1d. Build geographic dataset (Census + Election + PEN)

```python
# ── Census population ────────────────────────────────────────
census = census_raw[["NAME", "DP05_0001E"]].copy()
census.columns = ["state_name", "population"]
census["population"] = pd.to_numeric(census["population"], errors="coerce")
census = census.dropna(subset=["population"])
non_states = ["Puerto Rico", "District of Columbia", "Geographic Area Name"]
census = census[~census["state_name"].isin(non_states)]
print(f"Census: {len(census)} states")

# ── Election results (aggregate counties → states) ───────────
election = election_raw.groupby("STATE_NAME").agg(
    total_votes=("Votes_Tot", "sum"),
    trump_votes=("Votes_Trump", "sum"),
    harris_votes=("Votes_Harris", "sum"),
).reset_index()
election["trump_pct"] = election["trump_votes"] / election["total_votes"]
election["harris_pct"] = election["harris_votes"] / election["total_votes"]
election["political_lean"] = np.where(election["trump_pct"] > 0.5, "Republican", "Democrat")
election = election.rename(columns={"STATE_NAME": "state_name"})
print(f"Election: {len(election)} states")

# ── PEN ban counts by state ─────────────────────────────────
state_bans = pen_raw.groupby("State").agg(
    total_bans=("Title", "size"),
    unique_titles=("Title", "nunique"),
    unique_districts=("District", "nunique"),
    school_years=("School Year", "nunique"),
).reset_index().rename(columns={"State": "state_name"})

# Challenge reason counts by state
for cat_name in CATEGORY_KEYWORDS:
    col_safe = cat_name.lower().replace(" ", "_").replace("&", "and").replace("+", "plus")[:20]
    state_bans[f"has_{col_safe}"] = pen_raw.groupby("State")["categories_list"].apply(
        lambda lists: sum(1 for sublist in lists if cat_name in sublist)
    ).values

# ── Merge everything ─────────────────────────────────────────
geo = state_bans.merge(census, on="state_name", how="left")
geo = geo.merge(election, on="state_name", how="left")
geo["bans_per_100k"] = geo["total_bans"] / geo["population"] * 100_000

print(f"\nGeographic dataset: {len(geo)} states")
print(f"  With population: {geo['population'].notna().sum()}")
print(f"  With election data: {geo['trump_pct'].notna().sum()}")
print(f"  Columns: {list(geo.columns)}")
```

### 1e. Clean Goodreads

```python
gr = gr_raw[gr_raw["scrape_status"] == "ok"].copy()
gr = gr.drop_duplicates(subset=["book_title"], keep="first")

# Log transform for viz
gr["log_rating_count"] = np.log10(gr["rating_count"].clip(lower=1))

# Merge ban counts
gr = gr.merge(ban_counts, on="book_title", how="left")

print(f"Goodreads: {len(gr)} unique titles")
print(f"  Mean rating: {gr['avg_rating'].mean():.2f}")
print(f"  Median rating count: {gr['rating_count'].median():,.0f}")
print(f"  Has polarization: {gr['polarization'].notna().sum()}")
```

## 2. Statistical Tests

All 7 tests for the presentation.

### Test 1: One-Sample T-Test — Do bans affect search interest on average?

```python
finite = gt["pct_finite"].dropna()
t_stat, p_val = stats.ttest_1samp(finite, 0)
d = finite.mean() / finite.std()

print("TEST 1: One-Sample T-Test")
print(f"  H₀: Mean percent change = 0 (bans have no effect)")
print(f"  H₁: Mean percent change ≠ 0")
print(f"  N = {len(finite)}")
print(f"  Mean = {finite.mean():.2f}%")
print(f"  Median = {finite.median():.2f}%")
print(f"  SD = {finite.std():.2f}%")
print(f"  t = {t_stat:.4f}")
print(f"  p = {p_val:.6f}")
print(f"  Cohen's d = {d:.4f}")
print(f"  → {'SIGNIFICANT' if p_val < 0.05 else 'NOT significant'} at α=0.05")

# Proportion test
increased = gt["searches_increased"].sum()
total_obs = len(gt)
z = (increased / total_obs - 0.5) / np.sqrt(0.25 / total_obs)
p_prop = 2 * (1 - stats.norm.cdf(abs(z)))
print(f"\n  Proportion increased: {increased}/{total_obs} ({increased/total_obs*100:.1f}%)")
print(f"  z = {z:.4f}, p = {p_prop:.6f}")
```

### Test 2: Kruskal-Wallis — Streisand Effect by Challenge Reason Category

```python
# Explode so each book × reason is one row
exploded = cats_gt.explode("reasons_list")
exploded = exploded[exploded["reasons_list"].notna() & (exploded["reasons_list"] != "")]
exploded = exploded.rename(columns={"reasons_list": "reason"})
reasons = sorted(exploded["reason"].unique())

print("TEST 2: Kruskal-Wallis — By Challenge Reason")
print(f"\n  {'Category':<40s} {'N':>5s} {'Spike%':>7s} {'MeanΔ':>8s}")
print(f"  {'-'*40} {'-'*5} {'-'*7} {'-'*8}")
for reason in reasons:
    sub = exploded[exploded["reason"] == reason]
    n = len(sub)
    sr = sub["searches_increased"].mean() * 100
    md_ = sub["spike_magnitude"].mean()
    print(f"  {reason:<40s} {n:>5d} {sr:>6.1f}% {md_:>+7.2f}")

groups = [g["spike_magnitude"].dropna().values
          for _, g in exploded.groupby("reason") if len(g) >= 5]
groups = [g for g in groups if len(g) >= 5]

if len(groups) >= 2:
    h_stat, kw_p = stats.kruskal(*groups)
    n_total = sum(len(g) for g in groups)
    k = len(groups)
    eta2 = (h_stat - k + 1) / (n_total - k)
    print(f"\n  H = {h_stat:.4f}")
    print(f"  p = {kw_p:.6f}")
    print(f"  η² = {eta2:.4f}")
    print(f"  → {'SIGNIFICANT' if kw_p < 0.05 else 'NOT significant'} at α=0.05")
```

### Test 3: Post-Hoc Pairwise Mann-Whitney U (Bonferroni)

```python
reason_groups = {r: exploded[exploded["reason"] == r]["spike_magnitude"].dropna()
                 for r in reasons if len(exploded[exploded["reason"] == r]) >= 5}
pairs = list(combinations(reason_groups.keys(), 2))
n_pairs = len(pairs)
alpha_bonf = 0.05 / n_pairs if n_pairs > 0 else 0.05

print(f"TEST 3: Pairwise Mann-Whitney U")
print(f"  {n_pairs} pairs, Bonferroni α = {alpha_bonf:.6f}")

sig_pairs = 0
for r1, r2 in pairs:
    u_stat, u_p = stats.mannwhitneyu(reason_groups[r1], reason_groups[r2], alternative="two-sided")
    if u_p < alpha_bonf:
        print(f"  ✓ {r1} vs {r2}: p = {u_p:.6f}")
        sig_pairs += 1

if sig_pairs == 0:
    print("  No pairs significant after Bonferroni correction.")
print(f"  → {sig_pairs}/{n_pairs} significant")
```

### Test 4: Chi-Square — Spike vs Decline Rates by Category

```python
ct = pd.crosstab(exploded["reason"], exploded["searches_increased"])

print("TEST 4: Chi-Square — Spike Rate by Category")
if ct.shape[1] == 2:
    chi2, chi_p, dof, expected = stats.chi2_contingency(ct)
    n_chi = ct.values.sum()
    cramers_v = np.sqrt(chi2 / (n_chi * (min(ct.shape) - 1)))
    print(f"  χ² = {chi2:.4f}")
    print(f"  df = {dof}")
    print(f"  p = {chi_p:.6f}")
    print(f"  Cramér's V = {cramers_v:.4f}")
    print(f"  → {'SIGNIFICANT' if chi_p < 0.05 else 'NOT significant'}")
```

### Test 5: Kruskal-Wallis — By Pre-Ban Popularity Tier ⭐ (KEY FINDING)

```python
tier_order = ["Very Popular (50+)", "Popular (20-50)", "Moderate (5-20)",
              "Low Interest (<5)", "Niche (0)"]

print("TEST 5: Kruskal-Wallis — By Pre-Ban Popularity Tier")
print(f"\n  {'Tier':<25s} {'N':>5s} {'Spike%':>7s} {'MeanΔ':>8s} {'Mean%Chg':>10s}")
print(f"  {'-'*25} {'-'*5} {'-'*7} {'-'*8} {'-'*10}")

for tier in tier_order:
    sub = gt[gt["pre_ban_popularity"] == tier]
    if len(sub) == 0: continue
    n = len(sub)
    sr = sub["searches_increased"].mean() * 100
    md_ = sub["spike_magnitude"].mean()
    fs = sub["pct_finite"].dropna()
    mp = fs.mean() if len(fs) > 0 else np.nan
    ps = f"{mp:>+9.1f}%" if not np.isnan(mp) else "      N/A"
    print(f"  {tier:<25s} {n:>5d} {sr:>6.1f}% {md_:>+7.2f} {ps}")

tier_groups = [g["spike_magnitude"].dropna().values
               for _, g in gt.groupby("pre_ban_popularity") if len(g) >= 5]
tier_groups = [g for g in tier_groups if len(g) >= 5]

if len(tier_groups) >= 2:
    h5, p5 = stats.kruskal(*tier_groups)
    n5 = sum(len(g) for g in tier_groups)
    k5 = len(tier_groups)
    eta5 = (h5 - k5 + 1) / (n5 - k5)
    print(f"\n  H = {h5:.4f}")
    print(f"  p = {p5:.10f}")
    print(f"  η² = {eta5:.4f} ({'large' if eta5 > 0.14 else 'medium' if eta5 > 0.06 else 'small'})")
    print(f"  → {'SIGNIFICANT' if p5 < 0.05 else 'NOT significant'} at α=0.05")

# Post-hoc
print("\n  Post-hoc pairwise (Bonferroni):")
tier_dict = {t: gt[gt["pre_ban_popularity"] == t]["spike_magnitude"].dropna()
             for t in tier_order if len(gt[gt["pre_ban_popularity"] == t]) >= 5}
tier_pairs = list(combinations(tier_dict.keys(), 2))
alpha_t = 0.05 / len(tier_pairs) if tier_pairs else 0.05
sig = 0
for t1, t2 in tier_pairs:
    u, pu = stats.mannwhitneyu(tier_dict[t1], tier_dict[t2], alternative="two-sided")
    if pu < alpha_t:
        print(f"  ✓ {t1} vs {t2}: p = {pu:.8f}")
        sig += 1
print(f"  → {sig}/{len(tier_pairs)} significant")
```

### Test 6: Spearman Correlation — Ban Frequency vs Streisand Effect

```python
print("TEST 6: Spearman Correlation — Ban Count vs Spike")
for col in ["total_ban_instances", "states_banned_in"]:
    if col in gt.columns:
        valid = gt.dropna(subset=[col, "spike_magnitude"])
        r_s, p_s = stats.spearmanr(valid[col], valid["spike_magnitude"])
        print(f"\n  {col} vs spike_magnitude (N={len(valid)}):")
        print(f"    Spearman ρ = {r_s:.4f}, p = {p_s:.6f}")
        print(f"    → {'SIGNIFICANT' if p_s < 0.05 else 'NOT significant'}")
```

### Test 7: Political Correlation — Trump Vote % vs Bans Per Capita

```python
valid_pol = geo.dropna(subset=["trump_pct", "bans_per_100k"])

print("TEST 7: Political Correlation")
if len(valid_pol) >= 5:
    r_pol, p_pol = stats.spearmanr(valid_pol["trump_pct"], valid_pol["bans_per_100k"])
    print(f"  N = {len(valid_pol)} states")
    print(f"  Spearman ρ = {r_pol:.4f}, p = {p_pol:.6f}")
    print(f"  → {'SIGNIFICANT' if p_pol < 0.05 else 'NOT significant'}")

    red = valid_pol[valid_pol["political_lean"] == "Republican"]["bans_per_100k"]
    blue = valid_pol[valid_pol["political_lean"] == "Democrat"]["bans_per_100k"]
    if len(red) >= 2 and len(blue) >= 2:
        u_pol, p_mw = stats.mannwhitneyu(red, blue, alternative="two-sided")
        print(f"\n  Red states (N={len(red)}): mean = {red.mean():.1f}/100k")
        print(f"  Blue states (N={len(blue)}): mean = {blue.mean():.1f}/100k")
        print(f"  Mann-Whitney p = {p_mw:.6f}")
        print(f"  → {'SIGNIFICANT' if p_mw < 0.05 else 'NOT significant'}")
else:
    print("  ⚠ Not enough states with both Trump % and bans data")
```

### Chart 1: What Happens to Banned Books?

```python
counts = gt["streisand_category"].value_counts().reset_index()
counts.columns = ["Category", "Count"]
counts["Percent"] = (counts["Count"] / counts["Count"].sum() * 100).round(1)

cat_order = ["Strong Spike", "Mild Increase", "New Interest",
             "No Change", "Decline", "Strong Decline", "No Google Presence"]
cat_order = [c for c in cat_order if c in counts["Category"].values]

color_map = {
    "Strong Spike": TERRACOTTA, "Mild Increase": GOLD, "New Interest": CLAY,
    "No Change": SAGE, "Decline": CHARCOAL, "Strong Decline": "#6B6B6B",
    "No Google Presence": "#D0CCC0",
}

bars = (
    alt.Chart(counts).mark_bar(cornerRadiusTopLeft=4, cornerRadiusTopRight=4)
    .encode(
        x=alt.X("Category:N", sort=cat_order, title=None,
                 axis=alt.Axis(labelAngle=-30, labelFontSize=11)),
        y=alt.Y("Count:Q", title="Number of Books"),
        color=alt.Color("Category:N", sort=cat_order,
                         scale=alt.Scale(domain=list(color_map.keys()),
                                         range=list(color_map.values())),
                         legend=None),
        tooltip=["Category", "Count", "Percent"],
    )
    .properties(title={"text": "What Happens to Banned Books?",
                        "subtitle": "Distribution of Google Trends outcomes after book bans"})
)
text = (
    alt.Chart(counts).mark_text(dy=-10, fontSize=13, fontWeight="bold", color=TEXT_COLOR)
    .encode(x=alt.X("Category:N", sort=cat_order), y="Count:Q", text="Count:Q")
)
save_chart(bars + text, "01_outcome_distribution")
```

### Chart 2a: The Goldilocks Zone — Spike Rate ⭐

```python
tier_order = ["Very Popular (50+)", "Popular (20-50)", "Moderate (5-20)",
              "Low Interest (<5)", "Niche (0)"]
rows = []
for tier in tier_order:
    sub = gt[gt["pre_ban_popularity"] == tier]
    if len(sub) == 0: continue
    rows.append({
        "Tier": tier,
        "Spike Rate (%)": round(sub["searches_increased"].mean() * 100, 1),
        "Mean Spike Magnitude": round(sub["spike_magnitude"].mean(), 2),
        "N": len(sub),
    })
tier_df = pd.DataFrame(rows)

tier_colors = {"Very Popular (50+)": CHARCOAL, "Popular (20-50)": TERRACOTTA,
               "Moderate (5-20)": GOLD, "Low Interest (<5)": SAGE, "Niche (0)": CLAY}

bars = (
    alt.Chart(tier_df).mark_bar(cornerRadiusTopLeft=4, cornerRadiusTopRight=4)
    .encode(
        x=alt.X("Tier:N", sort=tier_order, title=None,
                 axis=alt.Axis(labelAngle=-20, labelFontSize=11)),
        y=alt.Y("Spike Rate (%):Q", title="% of Books with Increased Interest",
                 scale=alt.Scale(domain=[0, 105])),
        color=alt.Color("Tier:N", sort=tier_order,
                         scale=alt.Scale(domain=list(tier_colors.keys()),
                                         range=list(tier_colors.values())),
                         legend=None),
        tooltip=["Tier", "Spike Rate (%)", "N"],
    )
    .properties(title={"text": "The Goldilocks Zone: Who Benefits from Bans?",
                        "subtitle": "% of books that saw increased search interest after being banned"})
)
text = (
    alt.Chart(tier_df).mark_text(dy=-10, fontSize=13, fontWeight="bold", color=TEXT_COLOR)
    .encode(x=alt.X("Tier:N", sort=tier_order), y="Spike Rate (%):Q",
            text=alt.Text("Spike Rate (%):Q", format=".0f"))
)
save_chart(bars + text, "02a_goldilocks_spike_rate")
```

### Chart 5: Shifting Targets — Challenge Reasons Over Time

```python
categories_kw = {
    "LGBTQ+": ["lgbtq", "transgender", "homosexual", "gay", "lesbian", "queer", "nonbinary", "same-sex"],
    "Race & Inequality": ["racial", "racism", "slavery", "civil rights", "segregation", "discrimination"],
    "Sexual Content": ["sexual content", "sexually explicit", "sex scene", "nudity", "pornograph", "erotic"],
    "Violence": ["violence", "murder", "abuse", "assault", "gore", "torture", "shooting"],
    "Mental Health": ["mental health", "suicide", "depression", "self-harm", "trauma", "addiction"],
}

years = sorted(pen_raw["School Year"].unique())
rows = []
for year in years:
    sub = pen_raw[pen_raw["School Year"] == year]
    total = len(sub)
    desc = sub["Description"].fillna("")
    for cat_name, keywords in categories_kw.items():
        pattern = "|".join(keywords)
        count = desc.str.contains(pattern, case=False).sum()
        rows.append({"School Year": year, "Category": cat_name,
                      "Percentage": round(count / total * 100, 1) if total > 0 else 0})

shift_df = pd.DataFrame(rows)
cat_colors = {"LGBTQ+": TERRACOTTA, "Race & Inequality": GOLD, "Sexual Content": CHARCOAL,
              "Violence": CLAY, "Mental Health": SAGE}

chart = (
    alt.Chart(shift_df).mark_line(point=True, strokeWidth=2.5)
    .encode(
        x=alt.X("School Year:N", title=None, axis=alt.Axis(labelAngle=0)),
        y=alt.Y("Percentage:Q", title="% of Bans Citing This Reason"),
        color=alt.Color("Category:N",
                         scale=alt.Scale(domain=list(cat_colors.keys()),
                                         range=list(cat_colors.values())),
                         legend=alt.Legend(title="Challenge Reason")),
        strokeDash=alt.StrokeDash("Category:N", legend=None),
        tooltip=["School Year", "Category", "Percentage"],
    )
    .properties(title={"text": "The Targets Are Shifting",
                        "subtitle": "Challenge reason trends over time (classified from PEN Description text)"})
)
save_chart(chart, "05_shifting_targets", w=650, h=400)
```

### Chart 12: Challenge Reasons by Political Lean

```python
has_cols = [c for c in geo.columns if c.startswith("has_")]

rows = []
for lean in ["Republican", "Democrat"]:
    sub = geo[geo["political_lean"] == lean]
    total = sub["total_bans"].sum()
    if total == 0: continue
    for col in has_cols:
        reason = col.replace("has_", "").replace("_", " ").title()
        pct = sub[col].sum() / total * 100
        rows.append({"Political Lean": lean, "Challenge Reason": reason,
                      "Percentage": round(pct, 1)})

heat_df = pd.DataFrame(rows)

rects = (
    alt.Chart(heat_df).mark_rect(cornerRadius=4)
    .encode(
        x=alt.X("Political Lean:N", title=None),
        y=alt.Y("Challenge Reason:N", title=None),
        color=alt.Color("Percentage:Q",
                         scale=alt.Scale(domain=[0, heat_df["Percentage"].max()],
                                         range=["#F2E6D9", TERRACOTTA]),
                         legend=alt.Legend(title="% of Bans")),
        tooltip=["Political Lean", "Challenge Reason", "Percentage"],
    )
    .properties(title={"text": "Red and Blue States Ban Books for the Same Reasons",
                        "subtitle": "Challenge reason profiles by political lean"})
)
text = (
    alt.Chart(heat_df).mark_text(fontSize=13, fontWeight="bold", color=TEXT_COLOR)
    .encode(x="Political Lean:N", y="Challenge Reason:N",
            text=alt.Text("Percentage:Q", format=".1f"))
)
save_chart(rects + text, "12_reason_by_politics", w=400, h=400)
```
