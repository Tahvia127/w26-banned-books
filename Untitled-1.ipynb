```python
"""
Banned But Not Forgotten — Full Analysis Pipeline
====================================================
Run this script whenever you have new data. It processes everything
from raw inputs through cleaning, merging, statistical tests,
and geographic analysis, then exports all results.

INPUTS (put these in the same folder as this script):
  - PEN_Combined_21-25_Final.csv         (PEN America dataset)
  - google_trends_raw.csv                (Kevin's + new collection results, combined)
  - goodreads_current.csv                (Violet's Goodreads engagement data)
  - goodreads_ids_temp.csv               (Violet's Goodreads ID lookup)
  - full_book_details.csv                (Violet's expanded book details)
  - ACSDP5Y2023.DP05-Data.csv           (Census population data)
  - election_results.xlsx                (2024 Presidential Election, county-level)

OUTPUTS (all written to an 'outputs/' subfolder):
  - google_trends_cleaned.csv            (cleaned Google Trends analysis set)
  - goodreads_cleaned.csv                (cleaned Goodreads data)
  - title_categories.csv                 (challenge reason categories per title)
  - pen_categorized_full.csv             (full PEN dataset with categories)
  - categories_x_google_trends.csv       (merged categories + trends)
  - state_geographic_analysis.csv        (state-level geographic data)
  - master_dataset.csv                   (everything merged at title level)
  - statistical_results.txt              (all test results)

USAGE:
  python analysis_pipeline.py

  Or run individual sections:
  python analysis_pipeline.py --only clean_trends
  python analysis_pipeline.py --only clean_goodreads
  python analysis_pipeline.py --only categories
  python analysis_pipeline.py --only stats
  python analysis_pipeline.py --only geographic
  python analysis_pipeline.py --only merge
"""

import pandas as pd
import numpy as np
import re
import os
import sys
import zipfile
from datetime import datetime
from collections import Counter
from scipy import stats
from itertools import combinations

# ============================================================
# CONFIG — EDIT THESE PATHS IF YOUR FILES ARE ELSEWHERE
# ============================================================

PATHS = {
    'pen':              'PEN_Combined_21-25_Final.csv',
    'google_trends':    'google_trends_raw.csv',
    'goodreads':        'goodreads_current.csv',
    'goodreads_ids':    'goodreads_ids_temp.csv',
    'book_details':     'full_book_details.csv',
    'census':           'ACSDP5Y2023.DP05-Data.csv',
    'election':         'election_results.xlsx',  # or .zip path
}

OUTPUT_DIR = 'outputs'
os.makedirs(OUTPUT_DIR, exist_ok=True)


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def log(msg):
    print(f'[{datetime.now():%H:%M:%S}] {msg}')


def safe_load(key, required=True):
    """Load a CSV/Excel file, return None if not found and not required."""
    path = PATHS[key]
    if not os.path.exists(path):
        if required:
            print(f'ERROR: Required file not found: {path}')
            print(f'       Update PATHS["{key}"] at the top of this script.')
            sys.exit(1)
        else:
            log(f'Optional file not found, skipping: {path}')
            return None
    if path.endswith('.xlsx'):
        return pd.read_excel(path)
    return pd.read_csv(path)


def normalize_ban_date(val):
    """Convert PEN ban date strings to YYYY-MM-DD."""
    if pd.isna(val):
        return None
    val = re.sub(r'[\u00A0]', ' ', str(val).strip())
    for fmt in ['%B %Y', '%b %Y', '%B %y', '%b %y']:
        try:
            return datetime.strptime(val, fmt).strftime('%Y-%m-%d')
        except ValueError:
            pass
    m = re.match(r'^([A-Za-z]{3})-(\d{2})$', val)
    if m:
        try:
            return datetime.strptime(f'{m.group(1)} 20{m.group(2)}', '%b %Y').strftime('%Y-%m-%d')
        except ValueError:
            pass
    return None


# ============================================================
# SECTION 1: CLEAN GOOGLE TRENDS DATA
# ============================================================

def clean_google_trends():
    log('=== CLEANING GOOGLE TRENDS DATA ===')

    gt = safe_load('google_trends')
    pen = safe_load('pen')

    # Remove failed collections
    if 'collection_successful' in gt.columns:
        gt = gt[gt['collection_successful'] == True].copy()
    log(f'  Loaded {len(gt)} successful collections, {gt["book_title"].nunique()} unique titles')

    # Classify Streisand category
    def classify(row):
        before, after = row['avg_search_before'], row['avg_search_after']
        pct = row['percent_change']
        if before == 0 and after == 0:
            return 'No Google Presence'
        if before == 0 and after > 0:
            return 'New Interest'
        if np.isfinite(pct):
            if pct > 25:   return 'Strong Spike'
            elif pct > 0:  return 'Mild Increase'
            elif pct > -10: return 'No Change'
            elif pct > -50: return 'Decline'
            else:           return 'Strong Decline'
        return 'Unknown'

    gt['streisand_category'] = gt.apply(classify, axis=1)
    gt['pct_change_clean'] = gt['percent_change'].replace([np.inf, -np.inf], np.nan)

    # Popularity tier
    def pop_tier(avg):
        if avg == 0:   return 'Unknown/Niche'
        elif avg < 5:  return 'Low Interest'
        elif avg < 20: return 'Moderate Interest'
        elif avg < 50: return 'Popular'
        else:          return 'Very Popular'

    gt['pre_ban_popularity'] = gt['avg_search_before'].apply(pop_tier)
    gt['spike_magnitude'] = gt['avg_search_after'] - gt['avg_search_before']
    gt['searches_increased'] = gt['avg_search_after'] > gt['avg_search_before']

    # Merge PEN metadata
    pen_deduped = pen.drop_duplicates(subset='Title', keep='first')
    pen_lookup = pen_deduped[['Title', 'Author', 'Genre', 'Ban Status']].rename(
        columns={'Title': 'book_title', 'Author': 'author_pen', 'Genre': 'genre', 'Ban Status': 'ban_status'})

    ban_counts = pen.groupby('Title').agg(
        total_ban_instances=('Title', 'size'),
        states_banned_in=('State', 'nunique'),
        districts_banned_in=('District', 'nunique')
    ).reset_index().rename(columns={'Title': 'book_title'})

    gt = gt.merge(pen_lookup, on='book_title', how='left', suffixes=('', '_pen'))
    gt = gt.merge(ban_counts, on='book_title', how='left')

    # Split into full set and analysis set (excludes No Google Presence)
    analysis = gt[gt['streisand_category'] != 'No Google Presence'].copy()

    # Export
    cols = ['book_title', 'author', 'state', 'ban_date', 'genre', 'ban_status',
            'total_ban_instances', 'states_banned_in', 'districts_banned_in',
            'pre_ban_popularity', 'avg_search_before', 'avg_search_after',
            'spike_magnitude', 'max_search_before', 'max_search_after',
            'pct_change_clean', 'streisand_category', 'searches_increased', 'volatility']

    # Use only columns that exist
    cols = [c for c in cols if c in analysis.columns]
    analysis = analysis[cols].rename(columns={'pct_change_clean': 'percent_change'})
    analysis = analysis.sort_values('spike_magnitude', ascending=False).reset_index(drop=True)

    out_path = os.path.join(OUTPUT_DIR, 'google_trends_cleaned.csv')
    analysis.to_csv(out_path, index=False)

    # Print summary
    n = len(analysis)
    finite = analysis['percent_change'].dropna()
    log(f'  Analysis set: {n} rows, {analysis["book_title"].nunique()} unique titles')
    log(f'  Mean % change (finite): {finite.mean():.1f}%')
    log(f'  Searches increased: {analysis["searches_increased"].sum()} ({analysis["searches_increased"].mean()*100:.1f}%)')

    cats = analysis['streisand_category'].value_counts()
    for cat, count in cats.items():
        log(f'    {cat:20s}: {count:4d} ({count/n*100:.1f}%)')

    log(f'  Saved: {out_path}')
    return analysis


# ============================================================
# SECTION 2: CLEAN GOODREADS DATA
# ============================================================

def clean_goodreads():
    log('=== CLEANING GOODREADS DATA ===')

    gc = safe_load('goodreads', required=False)
    if gc is None:
        return None

    ids = safe_load('goodreads_ids', required=False)
    fb = safe_load('book_details', required=False)
    pen = safe_load('pen')

    # Fix study guide contamination
    gc['is_study_guide'] = gc['goodreads_url'].str.contains(
        'study-guide|Study.Guide|summary-study', na=False, regex=True)

    if ids is not None:
        ids_found = ids[ids['found'] == True].copy()
        ids_found['is_sg'] = ids_found['goodreads_url'].str.contains(
            'study-guide|Study.Guide|summary-study', na=False, regex=True)
        ids_correct = ids_found[~ids_found['is_sg']][
            ['book_title', 'goodreads_id', 'goodreads_url']
        ].drop_duplicates(subset='book_title', keep='first')

        for idx, row in gc[gc['is_study_guide']].iterrows():
            correct = ids_correct[ids_correct['book_title'] == row['book_title']]
            if len(correct) > 0:
                gc.at[idx, 'goodreads_id'] = correct.iloc[0]['goodreads_id']
                gc.at[idx, 'goodreads_url'] = correct.iloc[0]['goodreads_url']
                gc.at[idx, 'is_study_guide'] = False

    log(f'  Study guides remaining: {gc["is_study_guide"].sum()}')

    # Parse star distributions
    def parse_count(val):
        if pd.isna(val): return np.nan
        m = re.match(r'([\d,]+)', str(val).strip())
        return int(m.group(1).replace(',', '')) if m else np.nan

    def parse_pct(val):
        if pd.isna(val): return np.nan
        m = re.search(r'\((\d+)%\)', str(val))
        return int(m.group(1)) if m else np.nan

    for star in [5, 4, 3, 2, 1]:
        col = f'dist_{star}_star'
        if col in gc.columns:
            gc[f'count_{star}_star'] = gc[col].apply(parse_count)
            gc[f'pct_{star}_star'] = gc[col].apply(parse_pct)

    # Derived metrics
    gc['log_rating_count'] = np.log10(gc['rating_count'].clip(lower=1))
    gc['polarization'] = (gc.get('pct_5_star', pd.Series(0)).fillna(0) +
                          gc.get('pct_1_star', pd.Series(0)).fillna(0)) / 100
    gc['rating_skew'] = (gc.get('pct_5_star', pd.Series(0)).fillna(0) -
                         gc.get('pct_1_star', pd.Series(0)).fillna(0)) / 100

    # Quality flags
    gc['data_quality'] = 'Good'
    gc.loc[gc['is_study_guide'], 'data_quality'] = 'Study Guide'
    gc.loc[gc['rating_count'] < 10, 'data_quality'] = 'Very Low Ratings'
    gc.loc[(gc['rating_count'] >= 10) & (gc['rating_count'] < 100) &
           (gc['data_quality'] == 'Good'), 'data_quality'] = 'Low Ratings'

    # Goodreads genres from full_book_details
    if fb is not None:
        junk = 'Study Guide|Book Analysis|Bookclub-in-a-Box|CliffsNotes|SparkNotes'
        fb_clean = fb[~fb['title'].str.contains(junk, na=False, regex=True)]
        fb_genres = fb_clean[fb_clean['genres'].notna()].drop_duplicates(
            subset='book_id', keep='first')
        gc['goodreads_genres'] = gc['goodreads_id'].map(
            dict(zip(fb_genres['book_id'], fb_genres['genres'])))

    # Merge PEN
    pen_deduped = pen.drop_duplicates(subset='Title', keep='first')
    ban_counts = pen.groupby('Title').agg(
        total_ban_instances=('Title', 'size'),
        states_banned_in=('State', 'nunique')
    ).reset_index().rename(columns={'Title': 'book_title'})

    gc = gc.merge(pen_deduped[['Title', 'Genre']].rename(
        columns={'Title': 'book_title', 'Genre': 'pen_genre'}), on='book_title', how='left')
    gc = gc.merge(ban_counts, on='book_title', how='left')

    out_path = os.path.join(OUTPUT_DIR, 'goodreads_cleaned.csv')
    gc.to_csv(out_path, index=False)

    good = gc[gc['data_quality'] == 'Good']
    log(f'  Total: {len(gc)}, Good quality: {len(good)}')
    log(f'  Avg rating (good): {good["avg_rating"].mean():.2f}')
    log(f'  Saved: {out_path}')

    # Genre supplement
    if fb is not None:
        fb_clean['title_norm'] = fb_clean['title'].str.strip().str.lower()
        pen_titles = pen.drop_duplicates(subset='Title')
        pen_titles = pen_titles.copy()
        pen_titles['title_norm'] = pen_titles['Title'].str.strip().str.lower()
        matched = fb_clean.merge(pen_titles[['title_norm', 'Title']], on='title_norm', how='inner')
        genre_sup = matched[matched['genres'].notna()][['Title', 'genres', 'avg_rating', 'pages']].copy()
        genre_sup = genre_sup.rename(columns={'Title': 'book_title', 'genres': 'goodreads_genres'})
        genre_sup = genre_sup.drop_duplicates(subset='book_title', keep='first')
        sup_path = os.path.join(OUTPUT_DIR, 'goodreads_genre_supplement.csv')
        genre_sup.to_csv(sup_path, index=False)
        log(f'  Genre supplement: {len(genre_sup)} titles → {sup_path}')

    return gc


# ============================================================
# SECTION 3: CATEGORY MAPPING
# ============================================================

# --- The mapping dictionaries ---

GENRE_TO_REASON = {
    # LGBTQ+
    'LGBT': 'LGBTQ+ identity & sexuality', 'LGBTQ': 'LGBTQ+ identity & sexuality',
    'LGBTQ+': 'LGBTQ+ identity & sexuality', 'QUEER': 'LGBTQ+ identity & sexuality',
    'LESBIAN': 'LGBTQ+ identity & sexuality', 'GAY': 'LGBTQ+ identity & sexuality',
    'BISEXUAL': 'LGBTQ+ identity & sexuality', 'TRANSGENDER': 'LGBTQ+ identity & sexuality',
    'TRANS': 'LGBTQ+ identity & sexuality', 'NONBINARY': 'LGBTQ+ identity & sexuality',
    'DRAG': 'LGBTQ+ identity & sexuality', 'PRIDE': 'LGBTQ+ identity & sexuality',
    'GENDER IDENTITY': 'LGBTQ+ identity & sexuality', 'SAME-SEX': 'LGBTQ+ identity & sexuality',
    'HOMOSEXUALITY': 'LGBTQ+ identity & sexuality',

    # Race & systemic inequality
    'RACE': 'Race & systemic inequality', 'RACISM': 'Race & systemic inequality',
    'AFRICAN AMERICAN': 'Race & systemic inequality', 'BLACK': 'Race & systemic inequality',
    'BLACK LIVES MATTER': 'Race & systemic inequality', 'CIVIL RIGHTS': 'Race & systemic inequality',
    'SLAVERY': 'Race & systemic inequality', 'ANTIRACIST': 'Race & systemic inequality',
    'ANTI-RACIST': 'Race & systemic inequality', 'INDIGENOUS': 'Race & systemic inequality',
    'NATIVE AMERICAN': 'Race & systemic inequality', 'LATINO': 'Race & systemic inequality',
    'LATINA': 'Race & systemic inequality', 'HISPANIC': 'Race & systemic inequality',
    'MULTICULTURAL': 'Race & systemic inequality', 'CULTURAL': 'Race & systemic inequality',
    'DIVERSITY': 'Race & systemic inequality', 'SOCIAL JUSTICE': 'Race & systemic inequality',
    'COLONIALISM': 'Race & systemic inequality', 'APARTHEID': 'Race & systemic inequality',
    'IMMIGRATION': 'Race & systemic inequality', 'HOLOCAUST': 'Race & systemic inequality',

    # Sexual / romantic content
    'ROMANCE': 'Sexual or romantic content', 'EROTICA': 'Sexual or romantic content',
    'EROTIC': 'Sexual or romantic content', 'NEW ADULT': 'Sexual or romantic content',
    'CHICK LIT': 'Sexual or romantic content', 'STEAMY': 'Sexual or romantic content',
    'ADULT ROMANCE': 'Sexual or romantic content', 'DARK ROMANCE': 'Sexual or romantic content',
    'SEXUAL': 'Sexual or romantic content', 'SEXUALITY': 'Sexual or romantic content',
    'SEX EDUCATION': 'Sexual or romantic content', 'PUBERTY': 'Sexual or romantic content',

    # Gender politics
    'FEMINISM': 'Gender politics & feminism', 'FEMINIST': 'Gender politics & feminism',
    'WOMENS': 'Gender politics & feminism', "WOMEN'S STUDIES": 'Gender politics & feminism',
    'GENDER': 'Gender politics & feminism', 'GENDER STUDIES': 'Gender politics & feminism',
    'BODY IMAGE': 'Gender politics & feminism',

    # Mental health & trauma
    'MENTAL HEALTH': 'Mental health & trauma', 'MENTAL ILLNESS': 'Mental health & trauma',
    'DEPRESSION': 'Mental health & trauma', 'ANXIETY': 'Mental health & trauma',
    'SUICIDE': 'Mental health & trauma', 'SELF-HARM': 'Mental health & trauma',
    'TRAUMA': 'Mental health & trauma', 'ABUSE': 'Mental health & trauma',
    'ADDICTION': 'Mental health & trauma', 'EATING DISORDER': 'Mental health & trauma',
    'PTSD': 'Mental health & trauma', 'GRIEF': 'Mental health & trauma',
    'DEATH': 'Mental health & trauma', 'RECOVERY': 'Mental health & trauma',
    'DRUGS': 'Mental health & trauma', 'SUBSTANCE ABUSE': 'Mental health & trauma',

    # Violence & mature themes
    'DARK': 'Violence & mature themes', 'HORROR': 'Violence & mature themes',
    'VIOLENCE': 'Violence & mature themes', 'CRIME': 'Violence & mature themes',
    'MURDER': 'Violence & mature themes', 'WAR': 'Violence & mature themes',
    'DYSTOPIA': 'Violence & mature themes', 'DYSTOPIAN': 'Violence & mature themes',
    'THRILLER': 'Violence & mature themes', 'SUSPENSE': 'Violence & mature themes',
    'MYSTERY THRILLER': 'Violence & mature themes', 'TRUE CRIME': 'Violence & mature themes',
    'DARK FANTASY': 'Violence & mature themes', 'SURVIVAL': 'Violence & mature themes',

    # Religion & worldview
    'RELIGION': 'Religion & worldview', 'RELIGIOUS': 'Religion & worldview',
    'ATHEISM': 'Religion & worldview', 'WITCHCRAFT': 'Religion & worldview',
    'OCCULT': 'Religion & worldview', 'MAGIC': 'Religion & worldview',
    'SPIRITUALITY': 'Religion & worldview', 'SATANISM': 'Religion & worldview',

    # Age-appropriateness
    'YOUNG ADULT': 'Age-appropriateness concerns', 'YOUNG ADULT FICTION': 'Age-appropriateness concerns',
    'COMING OF AGE': 'Age-appropriateness concerns', 'TEEN': 'Age-appropriateness concerns',
    'JUVENILE FICTION': 'Age-appropriateness concerns', 'JUVENILE NONFICTION': 'Age-appropriateness concerns',
    'MIDDLE GRADE': 'Age-appropriateness concerns', 'CHILDRENS': 'Age-appropriateness concerns',
    'PICTURE BOOKS': 'Age-appropriateness concerns', 'KIDS': 'Age-appropriateness concerns',
}

DESCRIPTION_KEYWORDS = {
    'LGBTQ+ identity & sexuality': [
        r'\bgay\b', r'\blesbian\b', r'\bbisexual\b', r'\btransgender\b',
        r'\btrans\b', r'\bqueer\b', r'\bnon-?binary\b', r'\bsame[- ]sex\b',
        r'\blgbt', r'\bdrag queen\b', r'\btwo moms\b', r'\btwo dads\b',
        r'\bpronoun', r'\bgender identity\b', r'\bcoming out\b'],
    'Race & systemic inequality': [
        r'\bracis[mt]\b', r'\bslavery\b', r'\bsegregation\b', r'\bcivil rights\b',
        r'\bblack lives\b', r'\banti-?racist\b', r'\bjim crow\b',
        r'\bapartheid\b', r'\bcolonial', r'\bimmigra', r'\bholocaust\b',
        r'\bgenocide\b', r'\bindigenous\b', r'\bnative american\b'],
    'Sexual or romantic content': [
        r'\bsex\b', r'\bsexual\b', r'\bnaked\b', r'\bnudit', r'\beroti',
        r'\bintimacy\b', r'\bintimate\b', r'\bpubert', r'\breproduct',
        r'\bmenstruat', r'\bbodies\b.*\bchanging\b'],
    'Mental health & trauma': [
        r'\bsuicid', r'\bself[- ]harm\b', r'\bdepression\b', r'\banxiety\b',
        r'\bmental health\b', r'\babuse\b', r'\btrauma\b', r'\baddiction\b',
        r'\bdrug\b', r'\bsubstance\b', r'\beating disorder\b', r'\bbull[yi]'],
    'Violence & mature themes': [
        r'\bmurder\b', r'\bviolence\b', r'\bkill', r'\bgun\b', r'\bshoot',
        r'\bwar\b', r'\btorture\b', r'\bassault\b', r'\brape\b'],
}


def build_categories():
    log('=== BUILDING CHALLENGE REASON CATEGORIES ===')

    pen = safe_load('pen')

    # Load Goodreads genre supplement if available
    sup_path = os.path.join(OUTPUT_DIR, 'goodreads_genre_supplement.csv')
    gr_genres = pd.read_csv(sup_path) if os.path.exists(sup_path) else None

    gr_main_path = os.path.join(OUTPUT_DIR, 'goodreads_cleaned.csv')
    gr_main = pd.read_csv(gr_main_path) if os.path.exists(gr_main_path) else None

    # One row per title
    title_df = pen.drop_duplicates(subset='Title', keep='first')[
        ['Title', 'Genre', 'Description']].copy()
    title_df = title_df.rename(columns={'Title': 'book_title'})

    # Add Goodreads genres
    gr_all_parts = []
    if gr_genres is not None:
        gr_all_parts.append(gr_genres[['book_title', 'goodreads_genres']])
    if gr_main is not None and 'goodreads_genres' in gr_main.columns:
        gr_all_parts.append(
            gr_main[gr_main['goodreads_genres'].notna()][['book_title', 'goodreads_genres']])
    if gr_all_parts:
        gr_all = pd.concat(gr_all_parts).drop_duplicates(subset='book_title', keep='first')
        title_df = title_df.merge(gr_all, on='book_title', how='left')
    else:
        title_df['goodreads_genres'] = np.nan

    # Map genres to reasons
    def get_tokens(row):
        tokens = set()
        for field in ['Genre', 'goodreads_genres']:
            val = row.get(field)
            if pd.notna(val):
                for g in re.split(r'[;,/]', str(val)):
                    g = g.strip().upper()
                    if g:
                        tokens.add(g)
        return tokens

    def reasons_from_genres(tokens):
        return {GENRE_TO_REASON[t] for t in tokens if t in GENRE_TO_REASON}

    def reasons_from_description(desc):
        reasons = set()
        if pd.isna(desc):
            return reasons
        desc_lower = str(desc).lower()
        for reason, patterns in DESCRIPTION_KEYWORDS.items():
            for pattern in patterns:
                if re.search(pattern, desc_lower):
                    reasons.add(reason)
                    break
        return reasons

    title_df['_tokens'] = title_df.apply(get_tokens, axis=1)
    title_df['_r_genre'] = title_df['_tokens'].apply(reasons_from_genres)
    title_df['_r_desc'] = title_df['Description'].apply(reasons_from_description)
    title_df['challenge_reasons'] = title_df.apply(
        lambda r: sorted(r['_r_genre'] | r['_r_desc']), axis=1)
    title_df['num_reasons'] = title_df['challenge_reasons'].apply(len)
    title_df['primary_reason'] = title_df['challenge_reasons'].apply(
        lambda x: x[0] if x else 'Uncategorized')

    total = len(title_df)
    categorized = (title_df['num_reasons'] > 0).sum()
    log(f'  {categorized}/{total} titles categorized ({categorized/total*100:.1f}%)')

    reason_counts = Counter()
    for reasons in title_df['challenge_reasons']:
        for r in reasons:
            reason_counts[r] += 1
    for reason, count in reason_counts.most_common():
        log(f'    {reason:40s}: {count:5d} ({count/total*100:.1f}%)')

    # Export title-level categories
    cat_export = title_df[['book_title', 'challenge_reasons', 'num_reasons', 'primary_reason']].copy()
    cat_export['challenge_reasons'] = cat_export['challenge_reasons'].apply(
        lambda x: '; '.join(x) if x else '')
    cat_path = os.path.join(OUTPUT_DIR, 'title_categories.csv')
    cat_export.to_csv(cat_path, index=False)
    log(f'  Saved: {cat_path}')

    # Add categories to full PEN dataset
    cat_lookup = title_df[['book_title', 'challenge_reasons', 'num_reasons', 'primary_reason']].copy()
    pen_cat = pen.merge(
        cat_lookup.rename(columns={'book_title': 'Title'}), on='Title', how='left')
    pen_cat['challenge_reasons'] = pen_cat['challenge_reasons'].apply(
        lambda x: '; '.join(x) if isinstance(x, list) else '')

    pen_path = os.path.join(OUTPUT_DIR, 'pen_categorized_full.csv')
    pen_cat.to_csv(pen_path, index=False)
    log(f'  Saved: {pen_path}')

    return cat_export, pen_cat


# ============================================================
# SECTION 4: STATISTICAL TESTS
# ============================================================

def run_statistics():
    log('=== RUNNING STATISTICAL TESTS ===')

    gt_path = os.path.join(OUTPUT_DIR, 'google_trends_cleaned.csv')
    cat_path = os.path.join(OUTPUT_DIR, 'title_categories.csv')

    if not os.path.exists(gt_path) or not os.path.exists(cat_path):
        log('  ERROR: Need google_trends_cleaned.csv and title_categories.csv first.')
        log('  Run clean_trends and categories first.')
        return

    gt = pd.read_csv(gt_path)
    cats = pd.read_csv(cat_path)
    pen = safe_load('pen')

    # Merge categories into GT
    df = gt.merge(cats, on='book_title', how='left')

    # Add ban counts
    ban_counts = pen.groupby('Title').agg(
        total_ban_instances=('Title', 'size'),
        states_banned_in=('State', 'nunique')
    ).reset_index().rename(columns={'Title': 'book_title'})
    df = df.merge(ban_counts, on='book_title', how='left')

    # Also save the merged file
    merge_path = os.path.join(OUTPUT_DIR, 'categories_x_google_trends.csv')
    df.to_csv(merge_path, index=False)

    results = []
    results.append('STATISTICAL ANALYSIS RESULTS')
    results.append(f'Date: {datetime.now():%Y-%m-%d %H:%M}')
    results.append(f'N = {len(df)} titles with Google Trends + category data')
    results.append('=' * 65)

    # --- TEST 1: One-sample t-test ---
    finite = df['percent_change'].dropna()
    finite = finite[np.isfinite(finite)]
    t_stat, p_val = stats.ttest_1samp(finite, 0)
    d = finite.mean() / finite.std()

    results.append('')
    results.append('TEST 1: One-sample t-test (mean % change ≠ 0)')
    results.append(f'  N={len(finite)}, Mean={finite.mean():.2f}%, Median={finite.median():.2f}%')
    results.append(f'  t={t_stat:.4f}, p={p_val:.6f}, Cohen d={d:.4f}')
    results.append(f'  → {"SIGNIFICANT" if p_val < 0.05 else "NOT significant"} at α=0.05')

    # Proportion test
    increased = df['searches_increased'].sum()
    total = len(df)
    z = (increased / total - 0.5) / np.sqrt(0.25 / total)
    p_prop = 2 * (1 - stats.norm.cdf(abs(z)))
    results.append(f'  Proportion increased: {increased}/{total} ({increased/total*100:.1f}%), z={z:.4f}, p={p_prop:.6f}')

    # --- TEST 2: Kruskal-Wallis by challenge reason ---
    df['reasons_list'] = df['challenge_reasons'].apply(
        lambda x: [r.strip() for r in str(x).split(';') if r.strip()] if pd.notna(x) and str(x).strip() else [])
    exploded = df.explode('reasons_list').rename(columns={'reasons_list': 'reason'})
    exploded = exploded[exploded['reason'].notna() & (exploded['reason'] != '')]

    groups = {}
    for reason in exploded['reason'].unique():
        subset = exploded[exploded['reason'] == reason]['spike_magnitude'].dropna()
        if len(subset) >= 5:
            groups[reason] = subset

    results.append('')
    results.append('TEST 2: Kruskal-Wallis (Streisand by challenge reason)')
    if len(groups) >= 2:
        kw_stat, kw_p = stats.kruskal(*groups.values())
        all_vals = np.concatenate(list(groups.values()))
        gm = all_vals.mean()
        ss_b = sum(len(g) * (g.mean() - gm) ** 2 for g in groups.values())
        ss_t = sum((all_vals - gm) ** 2)
        eta_sq = ss_b / ss_t if ss_t > 0 else 0
        results.append(f'  Groups: {len(groups)}, H={kw_stat:.4f}, p={kw_p:.6f}, η²={eta_sq:.4f}')
        results.append(f'  → {"SIGNIFICANT" if kw_p < 0.05 else "NOT significant"}')
        for name, vals in sorted(groups.items(), key=lambda x: -x[1].mean()):
            results.append(f'    {name:42s}: N={len(vals):3d}, mean Δ={vals.mean():+6.2f}')
    else:
        results.append('  Not enough groups (need ≥2 with N≥5)')

    # --- TEST 3: Post-hoc pairwise ---
    results.append('')
    results.append('TEST 3: Pairwise Mann-Whitney U (post-hoc)')
    if len(groups) >= 2:
        pairs = list(combinations(groups.keys(), 2))
        bonf = 0.05 / max(len(pairs), 1)
        sig_pairs = []
        for a, b in pairs:
            u, p = stats.mannwhitneyu(groups[a], groups[b], alternative='two-sided')
            if p < bonf:
                sig_pairs.append((a, b, p))
        results.append(f'  {len(pairs)} comparisons, Bonferroni α={bonf:.5f}')
        if sig_pairs:
            for a, b, p in sorted(sig_pairs, key=lambda x: x[2]):
                results.append(f'  * {a[:30]:30s} vs {b[:30]:30s}: p={p:.6f}')
        else:
            results.append('  No pairs significant after correction')

    # --- TEST 4: Chi-square on spike rates ---
    results.append('')
    results.append('TEST 4: Chi-square (spike rates by challenge reason)')
    if len(groups) >= 2:
        cont_data = []
        for reason in groups:
            sub = exploded[exploded['reason'] == reason]
            spiked = int(sub['searches_increased'].sum())
            declined = len(sub) - spiked
            rate = spiked / (spiked + declined) * 100
            cont_data.append({'reason': reason, 'spiked': spiked, 'declined': declined, 'rate': rate})

        cont_df = pd.DataFrame(cont_data).set_index('reason')
        chi2, chi_p, dof, _ = stats.chi2_contingency(cont_df[['spiked', 'declined']])
        n_chi = cont_df[['spiked', 'declined']].sum().sum()
        k = min(cont_df.shape[0], 2)
        v = np.sqrt(chi2 / (n_chi * (k - 1))) if n_chi > 0 else 0
        results.append(f'  χ²={chi2:.4f}, df={dof}, p={chi_p:.6f}, Cramér V={v:.4f}')
        results.append(f'  → {"SIGNIFICANT" if chi_p < 0.05 else "NOT significant"}')
        for idx, row in cont_df.sort_values('rate', ascending=False).iterrows():
            results.append(f'    {idx:42s}: {row["rate"]:.1f}% spike rate')

    # --- TEST 5: Kruskal-Wallis by popularity tier ---
    results.append('')
    results.append('★ TEST 5: Kruskal-Wallis (Streisand by popularity tier)')
    pop_groups = {}
    for tier in df['pre_ban_popularity'].dropna().unique():
        subset = df[df['pre_ban_popularity'] == tier]['spike_magnitude'].dropna()
        if len(subset) >= 5:
            pop_groups[tier] = subset

    if len(pop_groups) >= 2:
        kw2, kw2_p = stats.kruskal(*pop_groups.values())
        all_pop = np.concatenate(list(pop_groups.values()))
        gm2 = all_pop.mean()
        ss_b2 = sum(len(g) * (g.mean() - gm2) ** 2 for g in pop_groups.values())
        ss_t2 = sum((all_pop - gm2) ** 2)
        eta2 = ss_b2 / ss_t2 if ss_t2 > 0 else 0
        results.append(f'  H={kw2:.4f}, p={kw2_p:.10f}, η²={eta2:.4f}')
        results.append(f'  → {"SIGNIFICANT" if kw2_p < 0.05 else "NOT significant"}')
        for name, vals in sorted(pop_groups.items(), key=lambda x: -x[1].mean()):
            spike_rate = (vals > 0).mean() * 100
            results.append(f'    {name:20s}: N={len(vals):3d}, mean Δ={vals.mean():+6.2f}, spike rate={spike_rate:.0f}%')

        # Post-hoc
        if kw2_p < 0.05:
            pop_pairs = list(combinations(pop_groups.keys(), 2))
            bonf2 = 0.05 / max(len(pop_pairs), 1)
            results.append(f'  Post-hoc (Bonferroni α={bonf2:.5f}):')
            for a, b in pop_pairs:
                u, p = stats.mannwhitneyu(pop_groups[a], pop_groups[b], alternative='two-sided')
                if p < bonf2:
                    results.append(f'    * {a:20s} vs {b:20s}: p={p:.6f}')

    # --- TEST 6: Correlation ---
    results.append('')
    results.append('TEST 6: Correlation (ban frequency vs Streisand)')

    # Add ban counts if not already present
    if 'total_ban_instances' not in df.columns:
        pen_raw = safe_load('pen')
        bc = pen_raw.groupby('Title').agg(
            total_ban_instances=('Title', 'size'),
            states_banned_in=('State', 'nunique')
        ).reset_index().rename(columns={'Title': 'book_title'})
        df = df.merge(bc, on='book_title', how='left')

    valid = df.dropna(subset=['total_ban_instances', 'spike_magnitude'])
    if len(valid) >= 10:
        r_s, p_s = stats.spearmanr(valid['total_ban_instances'], valid['spike_magnitude'])
        results.append(f'  Ban instances vs spike: ρ={r_s:.4f}, p={p_s:.6f}')
        r_st, p_st = stats.spearmanr(valid['states_banned_in'], valid['spike_magnitude'])
        results.append(f'  States banned vs spike: ρ={r_st:.4f}, p={p_st:.6f}')

    # Write results
    results_text = '\n'.join(results)
    print(results_text)

    out_path = os.path.join(OUTPUT_DIR, 'statistical_results.txt')
    with open(out_path, 'w') as f:
        f.write(results_text)
    log(f'  Saved: {out_path}')


# ============================================================
# SECTION 5: GEOGRAPHIC ANALYSIS
# ============================================================

def run_geographic():
    log('=== RUNNING GEOGRAPHIC ANALYSIS ===')

    pen_path = os.path.join(OUTPUT_DIR, 'pen_categorized_full.csv')
    if not os.path.exists(pen_path):
        log('  ERROR: Need pen_categorized_full.csv. Run categories first.')
        return

    pen = pd.read_csv(pen_path)
    gt_path = os.path.join(OUTPUT_DIR, 'google_trends_cleaned.csv')
    gt = pd.read_csv(gt_path) if os.path.exists(gt_path) else None

    # Census
    census = safe_load('census', required=False)
    if census is not None:
        census = pd.read_csv(PATHS['census'], skiprows=[1])
        census = census[['NAME', 'DP05_0001E']].copy()
        census.columns = ['state_name', 'population']
        census['population'] = pd.to_numeric(census['population'], errors='coerce')
        census = census.dropna(subset=['population'])
        non_states = ['Puerto Rico', 'District of Columbia', 'Geographic Area Name']
        census = census[~census['state_name'].isin(non_states)]
    else:
        census = pd.DataFrame(columns=['state_name', 'population'])

    # Election
    election_path = PATHS['election']
    election = pd.DataFrame()
    if os.path.exists(election_path):
        try:
            if election_path.endswith('.zip'):
                with zipfile.ZipFile(election_path) as z:
                    xlsx = [n for n in z.namelist() if 'Results' in n and n.endswith('.xlsx')][0]
                    election_raw = pd.read_excel(z.open(xlsx))
            else:
                election_raw = pd.read_excel(election_path)

            election = election_raw.groupby('STATE_NAME').agg(
                total_votes=('Votes_Tot', 'sum'),
                trump_votes=('Votes_Trump', 'sum'),
                harris_votes=('Votes_Harris', 'sum')
            ).reset_index()
            election['trump_pct'] = election['trump_votes'] / election['total_votes']
            election['harris_pct'] = election['harris_votes'] / election['total_votes']
            election['political_lean'] = np.where(election['trump_pct'] > 0.5, 'Republican', 'Democrat')
            election = election.rename(columns={'STATE_NAME': 'state_name'})
        except Exception as e:
            log(f'  Warning: could not load election data: {e}')

    # Build state-level stats
    reason_keys = ['LGBTQ', 'Race', 'Sexual', 'Violence', 'Mental', 'Religion', 'Age']
    agg_dict = {
        'total_bans': ('Title', 'size'),
        'unique_titles_banned': ('Title', 'nunique'),
        'unique_districts': ('District', 'nunique'),
        'school_years': ('School Year', 'nunique'),
    }
    for key in reason_keys:
        agg_dict[f'has_{key.lower()}'] = (
            'challenge_reasons',
            lambda x, k=key: x.str.contains(k, na=False).sum()
        )

    state_df = pen.groupby('State').agg(**agg_dict).reset_index()
    state_df = state_df.rename(columns={'State': 'state_name'})

    # Merge census + election
    state_df = state_df.merge(census, on='state_name', how='left')
    state_df['bans_per_100k'] = state_df['total_bans'] / state_df['population'] * 100000
    state_df['titles_per_100k'] = state_df['unique_titles_banned'] / state_df['population'] * 100000

    if len(election) > 0:
        state_df = state_df.merge(
            election[['state_name', 'trump_pct', 'harris_pct', 'political_lean']],
            on='state_name', how='left')

    # Challenge reason percentages
    for key in reason_keys:
        col = f'has_{key.lower()}'
        if col in state_df.columns:
            state_df[f'pct_{key.lower()}'] = state_df[col] / state_df['total_bans'] * 100

    out_path = os.path.join(OUTPUT_DIR, 'state_geographic_analysis.csv')
    state_df.to_csv(out_path, index=False)
    log(f'  Saved: {out_path} ({len(state_df)} states)')

    # Print top states
    log('  Top 10 by total bans:')
    for _, r in state_df.nlargest(10, 'total_bans').iterrows():
        pc = f"{r['bans_per_100k']:.1f}" if pd.notna(r.get('bans_per_100k')) else 'N/A'
        lean = r.get('political_lean', '?')
        lean = lean if pd.notna(lean) else '?'
        log(f"    {r['state_name']:20s}: {r['total_bans']:5d} bans, {pc:>6s}/100k, {lean}")

    # Political correlation
    valid = state_df.dropna(subset=['trump_pct', 'bans_per_100k'])
    if len(valid) >= 10:
        r_pol, p_pol = stats.spearmanr(valid['trump_pct'], valid['bans_per_100k'])
        log(f'  Political correlation: ρ={r_pol:.4f}, p={p_pol:.6f}')

        red = valid[valid['political_lean'] == 'Republican']['bans_per_100k']
        blue = valid[valid['political_lean'] == 'Democrat']['bans_per_100k']
        if len(red) >= 3 and len(blue) >= 3:
            _, mw_p = stats.mannwhitneyu(red, blue, alternative='two-sided')
            log(f'  Red mean={red.mean():.1f}/100k vs Blue mean={blue.mean():.1f}/100k, p={mw_p:.6f}')

    # Time trends
    log('  Yearly totals:')
    yearly = pen.groupby('School Year').size()
    for yr, n in yearly.sort_index().items():
        log(f'    {yr}: {n:,d}')

    # Streisand by state
    if gt is not None and 'state' in gt.columns:
        gt_state = gt.groupby('state').agg(
            n_books=('book_title', 'size'),
            mean_spike=('spike_magnitude', 'mean'),
            spike_rate=('searches_increased', 'mean')
        ).reset_index().rename(columns={'state': 'state_name'})
        gt_state = gt_state[gt_state['n_books'] >= 5]

        if len(gt_state) > 0:
            se_path = os.path.join(OUTPUT_DIR, 'streisand_by_state.csv')
            gt_state.to_csv(se_path, index=False)
            log(f'  Streisand by state ({len(gt_state)} states):')
            for _, r in gt_state.sort_values('mean_spike', ascending=False).iterrows():
                log(f"    {r['state_name']:20s}: N={r['n_books']:3.0f}, spike rate={r['spike_rate']*100:.1f}%")


# ============================================================
# SECTION 6: MASTER MERGED DATASET
# ============================================================

def build_master():
    log('=== BUILDING MASTER DATASET ===')

    pen = safe_load('pen')
    title_base = pen.drop_duplicates(subset='Title', keep='first')[['Title', 'Author', 'Genre']].copy()
    title_base = title_base.rename(columns={'Title': 'book_title', 'Author': 'author', 'Genre': 'pen_genre'})

    ban_counts = pen.groupby('Title').agg(
        total_ban_instances=('Title', 'size'),
        states_banned_in=('State', 'nunique'),
        districts_banned_in=('District', 'nunique'),
        first_school_year=('School Year', 'min'),
        last_school_year=('School Year', 'max')
    ).reset_index().rename(columns={'Title': 'book_title'})
    master = title_base.merge(ban_counts, on='book_title', how='left')

    # Categories
    cat_path = os.path.join(OUTPUT_DIR, 'title_categories.csv')
    if os.path.exists(cat_path):
        cats = pd.read_csv(cat_path)
        master = master.merge(cats, on='book_title', how='left')

    # Google Trends
    gt_path = os.path.join(OUTPUT_DIR, 'google_trends_cleaned.csv')
    if os.path.exists(gt_path):
        gt = pd.read_csv(gt_path)
        gt_deduped = gt.drop_duplicates(subset='book_title', keep='first')
        gt_cols = ['book_title', 'avg_search_before', 'avg_search_after', 'spike_magnitude',
                   'percent_change', 'streisand_category', 'searches_increased', 'pre_ban_popularity']
        gt_cols = [c for c in gt_cols if c in gt_deduped.columns]
        master = master.merge(gt_deduped[gt_cols], on='book_title', how='left')

    # Goodreads
    gr_path = os.path.join(OUTPUT_DIR, 'goodreads_cleaned.csv')
    if os.path.exists(gr_path):
        gr = pd.read_csv(gr_path)
        gr_cols = ['book_title', 'avg_rating', 'rating_count', 'polarization',
                   'rating_skew', 'goodreads_genres', 'data_quality']
        gr_cols = [c for c in gr_cols if c in gr.columns]
        master = master.merge(gr[gr_cols], on='book_title', how='left')

    out_path = os.path.join(OUTPUT_DIR, 'master_dataset.csv')
    master.to_csv(out_path, index=False)

    has_gt = master['streisand_category'].notna().sum() if 'streisand_category' in master.columns else 0
    has_gr = master['avg_rating'].notna().sum() if 'avg_rating' in master.columns else 0
    has_cat = (master['num_reasons'] > 0).sum() if 'num_reasons' in master.columns else 0

    log(f'  Master dataset: {len(master)} titles')
    log(f'    With Google Trends: {has_gt}')
    log(f'    With Goodreads: {has_gr}')
    log(f'    With categories: {has_cat}')
    log(f'  Saved: {out_path}')


# ============================================================
# MAIN
# ============================================================

SECTIONS = {
    'clean_trends': clean_google_trends,
    'clean_goodreads': clean_goodreads,
    'categories': build_categories,
    'stats': run_statistics,
    'geographic': run_geographic,
    'merge': build_master,
}

def main():
    if len(sys.argv) > 1 and sys.argv[1] == '--only':
        section = sys.argv[2] if len(sys.argv) > 2 else None
        if section in SECTIONS:
            SECTIONS[section]()
        else:
            print(f'Unknown section: {section}')
            print(f'Options: {", ".join(SECTIONS.keys())}')
    else:
        log('Running full pipeline...')
        log('')
        clean_google_trends()
        log('')
        clean_goodreads()
        log('')
        build_categories()
        log('')
        run_statistics()
        log('')
        run_geographic()
        log('')
        build_master()
        log('')
        log('DONE. All outputs in: ' + OUTPUT_DIR)


if __name__ == '__main__':
    main()
```
