```python
import pandas as pd

pen_path = "PEN_Combined_21-25_Final.csv"
pen = pd.read_csv(pen_path)

# Adjust these column names if yours differ:
books = (
    pen[["Title", "Author"]]
    .dropna(subset=["Title"])
    .drop_duplicates()
    .rename(columns={"Title": "book_title", "Author": "author"})
)

books.to_csv("pen_unique_books.csv", index=False)
print("Unique books:", len(books))
```
